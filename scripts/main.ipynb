```python
# plotting the per model behaviour of Pearson COrrelation when trained with 
# different training data sizes and how does different seeding changes their
# Correlation coefficient.

# load dataset
from pathlib import Path
import numpy as np
import pandas as pd

root = Path.cwd().parent
convergence_file = root / "results" / "convergence.csv"

# loading the dataset
try:
    df = pd.read_csv(convergence_file, encoding='utf-8')
    print(f"[x]    Data loading succesfull.")
except Exception as e:
    print(f"[x]    ERROR: {e}")

# creating the data series
train_size  =   df.iloc[:, 0].to_numpy()
std_r       =   df.iloc[:, 2].to_numpy()
mean_r      =   df.iloc[:, 1].to_numpy()

print(train_size)
print(std_r)
print(mean_r)
```

```python

## plotting the data

import matplotlib.pyplot as plt
plt.rcParams["font.family"] = "AnthropicSerif"

fig, ax = plt.subplots()
fig.set_figheight(3)
fig.set_figwidth(4)

ax.scatter(train_size, mean_r, s=(std_r*10000), c="#BB5566", zorder=2)

ax.set_title("Correlation shift of Mean Error VS Model disagreement")
ax.set_xlabel("Training Size")
ax.set_ylabel("Pearson Coefficient")

ax.set_xticks([300, 600, 900, 1200, 1500])
ax.set_yticks([0.3, 0.35, 0.4, 0.45, 0.5])

ax.grid(zorder=0)

file_path = root / "plots" / f"Correlation-changes-drift-withdatasize-and-variance-shift.png"
fig.tight_layout(pad=2.0)
fig.savefig(file_path, bbox_inches='tight', pad_inches=0.1)
```
